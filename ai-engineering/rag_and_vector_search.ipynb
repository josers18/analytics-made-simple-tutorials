# Analytics Made Simple — Semantic Vector Search & RAG from Scratch
Understand how embeddings, cosine distance, and Retrieval-Augmented Generation (RAG) work under the hood using pure Python.


```python
import math
import pandas as pd

def cosine_similarity(v1, v2):
    dot = sum(a * b for a, b in zip(v1, v2))
    norm_a = math.sqrt(sum(a * a for a in v1))
    norm_b = math.sqrt(sum(b * b for b in v2))
    return dot / (norm_a * norm_b)

# Toy knowledge base
kb = [
    {"doc_id": 101, "topic": "SQL Joins", "embedding": [0.95, 0.10, 0.05]},
    {"doc_id": 102, "topic": "Pandas Filtering", "embedding": [0.85, 0.40, 0.10]},
    {"doc_id": 103, "topic": "Customer Churn", "embedding": [0.12, 0.90, 0.40]}
]

query = [0.90, 0.20, 0.05] # Query about relational databases

scores = []
for item in kb:
    score = cosine_similarity(query, item["embedding"])
    scores.append({"doc_id": item["doc_id"], "topic": item["topic"], "similarity": round(score, 4)})

pd.DataFrame(scores).sort_values("similarity", ascending=False)
```
