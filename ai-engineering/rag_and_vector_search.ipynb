# Cosine similarity on three hand-written vectors

These numbers were typed by hand. This notebook does not embed text and it does not generate an answer. The SQLite version is [`03_vector_search_sqlite.py`](./03_vector_search_sqlite.py).

```python
import math

def cosine_similarity(v1, v2):
    dot = sum(a * b for a, b in zip(v1, v2))
    norm_a = math.sqrt(sum(a * a for a in v1))
    norm_b = math.sqrt(sum(b * b for b in v2))
    if norm_a == 0.0 or norm_b == 0.0:
        return 0.0
    return dot / (norm_a * norm_b)

kb = [
    {"doc_id": 101, "topic": "SQL Joins", "embedding": [0.95, 0.10, 0.05]},
    {"doc_id": 102, "topic": "Pandas Filtering", "embedding": [0.85, 0.40, 0.10]},
    {"doc_id": 103, "topic": "Customer Churn", "embedding": [0.12, 0.90, 0.40]},
]
query = [0.90, 0.20, 0.05]
scores = []
for item in kb:
    score = cosine_similarity(query, item["embedding"])
    scores.append((round(score, 4), item["topic"]))
for score, topic in sorted(scores, reverse=True):
    print(f"{score:.4f}  {topic}")
```
