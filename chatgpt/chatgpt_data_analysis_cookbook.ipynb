```python
import pandas as pd
import numpy as np

# Sample business dataset
df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer": ["Acme Corp", "Apex Logistics", "Beacon Health", "Crestview", "Delta Dyn"],
    "region": ["NA", "EU", "NA", "APAC", "EU"],
    "revenue": [1450.0, 3100.0, 450.0, 1980.0, 820.0],
    "status": ["completed", "completed", "pending", "completed", "completed"]
})

# Analysis
summary = df[df["status"] == "completed"].groupby("region")["revenue"].agg(["count", "sum"]).reset_index()
summary.columns = ["Region", "Completed Orders", "Total Revenue"]
summary
```
