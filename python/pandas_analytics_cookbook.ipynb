# Analytics Made Simple — Pandas Analytics Cookbook
An end-to-end interactive tutorial covering data loading, filtering, categorical standardization, grain verification, and Parquet export.


```python
import pandas as pd
import numpy as np
import re

print(f"Pandas version: {pd.__version__}")
```

### Step 1: Load and Inspect Operational Data


```python
df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer": ["Acme Corp", "Acme Corp.", "Apex Logistics Ltd", "Beacon Health", "Crestview"],
    "region": ["NA", "North America", "EU", "NA", "APAC"],
    "revenue": [1450.0, 820.0, 3100.0, 450.0, 1980.0],
    "status": ["COMPLETED", "completed", "COMPLETED", "pending", "completed"]
})
df_orders
```

### Step 2: Clean and Standardize Categories


```python
# Standardize region codes
region_map = {"NA": "North America", "EU": "Europe", "APAC": "Asia-Pacific"}
df_orders["region_clean"] = df_orders["region"].replace(region_map)

# Standardize status
df_orders["status_clean"] = df_orders["status"].str.strip().str.lower()

df_orders[["order_id", "customer", "region_clean", "revenue", "status_clean"]]
```

### Step 3: Grain Verification (Assertion)


```python
def assert_grain(df, key):
    assert len(df) == len(df.drop_duplicates(subset=[key])), f"Duplicate key detected in {key}!"
    print(f"✅ Verified: {key} is unique across all {len(df)} rows.")

assert_grain(df_orders, "order_id")
```
